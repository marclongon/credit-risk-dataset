```python
def bar_plot(col_alvo = None, d1 = None, viz_d1 = None, d2 = None, viz_d2 = None, w = "loan_amnt"):
  nvizd1 = viz_d1
  nvizd2 = viz_d2

  #Configura avaliacao 1 (d1)
  if viz_d1 == "qtd":
    viz_d1 = (
      df.groupby(col_alvo)[d1]
      .count().sort_index().rename("Quantidade")
    )
  elif viz_d1 == "volume":  
    viz_d1 = (
      df.groupby(col_alvo)[d1]
      .sum().sort_index().rename("Volume")
    )
  elif viz_d1 == "media":  
    viz_d1 = (
      df.groupby(col_alvo)[d1]
      .mean().sort_index().rename("Média")
    )
  elif viz_d1 == "media_pond": 
    viz_d1 = (
      df.groupby(col_alvo)
        .apply(
          lambda g: (g[w] * g[d1]).sum() / g[w].sum(),
          include_groups=False
          )
        .rename("Média Ponderada")
    )

  #Configura avaliacao 2 (d2)
  if viz_d2 == "qtd":
    viz_d2 = (
      df.groupby(col_alvo)[d2]
      .count().sort_index().rename("Quantidade")
    )
  elif viz_d2 == "volume":  
    viz_d2 = (
      df.groupby(col_alvo)[d2]
      .sum().sort_index().rename("Volume")
    )
  elif viz_d2 == "media":  
    viz_d2 = (
      df.groupby(col_alvo)[d2]
      .mean().sort_index().rename("Média")
    )
  elif viz_d2 == "media_pond": 
    viz_d2 = (
      df.groupby(col_alvo)
        .apply(
          lambda g: (g[w] * g[d2]).sum() / g[w].sum(),
          include_groups=False
          )
        .rename("Média Ponderada")
    )

  inad_comp = df.groupby(col_alvo).agg(
      qtd_inadimplente=("loan_status", lambda x: x.eq(1).sum()),
      volume_inadimplente=("loan_amnt", lambda x: x[df.loc[x.index, "loan_status"].eq(1)].sum())
  )

  inad_comp = inad_comp.rename(columns={
    "qtd_inadimplente": "Qtd Inadimplente",
    "volume_inadimplente": "Volume Inadimplente"
  }).rename_axis(col_alvo)
    
  de_para_col = {
    "person_age": "Idade",
    "person_income": "Renda",
    "person_home_ownership": "Moradia",
    "person_emp_length": "Prazo",
    "loan_intent": "Modalidade de Crédito",
    "loan_grade": "Classificação de Risco",
    "loan_amnt": "Valor",
    "loan_int_rate": "Taxa",
    "loan_status": "Status",
    "loan_percent_income": "Comprometimento",
    "cb_person_default_on_file": "Restritivos",
    "cb_person_cred_hist_length": "Histórico de Crédito",
  }

  de_para_viz = {
    "qtd": "Quantidade",
    "volume": "Volume",
    "media": "Média",
    "media_pond": "Média Ponderada"
  }

  l_col_alvo = de_para_col.get(col_alvo, col_alvo)
  l_d1 = de_para_col.get(d1, d1)
  l_d2 = de_para_col.get(d2, d2)
  l_viz_d1 = de_para_viz.get(nvizd1, nvizd1)
  l_viz_d2 = de_para_viz.get(nvizd2, nvizd2)

  #Une dataframes calculados para resumo dos dados
  # resumo = viz_d1.to_frame().merge(viz_d2.to_frame(), left_index=True, right_index=True)
  resumo_graf =  pd.concat(
    [viz_d1, viz_d2, inad_comp],
    axis=1
  )

  #Cria o grafico com 2 eixos (d1 e d2 - quando nao houver d2 nao traz dados)
  fig, ax1 = plt.subplots(figsize=(10, 5))
  ax1.bar(viz_d1.index, viz_d1.values, color="steelblue", alpha=0.7, label = l_d1)
  ax1.set_ylabel(l_d1)
  ax1.set_xlabel(l_col_alvo)
  ax1.set_title(f"Analise de {l_col_alvo} | {l_d1} ({l_viz_d1}) x {l_d2} ({l_viz_d2})") #
  ax1.legend(loc="upper left")

  if d2 is not None and viz_d2 is not None:
    ax2 = ax1.twinx()
    ax2.plot(viz_d2.index, viz_d2.values, color="orange", marker="o", label = l_d2)
    ax2.set_ylabel(l_d2)
    ax2.legend(loc="upper right")
  
  plt.show()
  
  return fig, print(resumo_graf)
```
